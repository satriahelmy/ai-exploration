```python
import pandas as pd

df = pd.read_csv("../data/telco_customer_churn.csv")

print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
display(df.dtypes)

print("\nDataset info:")
df.info()

print("\nTarget distribution:")
display(df["Churn"].value_counts())

print("\nTarget distribution (%):")
display(df["Churn"].value_counts(normalize=True).mul(100).round(2))

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))
```

```python
# Create an analysis copy
eda_df = df.copy()

# Convert blank TotalCharges values to missing values
eda_df["TotalCharges"] = pd.to_numeric(
    eda_df["TotalCharges"],
    errors="coerce"
)

eda_df["TotalCharges"].isna().sum()
```

```python
eda_df[eda_df["TotalCharges"].isna()][
    ["tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```python
import matplotlib.pyplot as plt

numeric_features = ["tenure", "MonthlyCharges", "TotalCharges"]

for col in numeric_features:
    eda_df.boxplot(column=col, by="Churn")
    plt.title(f"{col} by Churn")
    plt.suptitle("")
    plt.xlabel("Churn")
    plt.ylabel(col)
    plt.show()
```

```python
eda_df.groupby("Churn")[numeric_features].median()
```

```python
categorical_features = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod",
]

for col in categorical_features:
    churn_rate = (
        eda_df.groupby(col)["Churn"]
        .apply(lambda x: (x == "Yes").mean())
        .sort_values(ascending=False)
    )

    print(f"\n{col}")
    print((churn_rate * 100).round(2))
```

```python
features_to_plot = [
    "Contract",
    "InternetService",
    "PaymentMethod",
    "TechSupport",
]

for col in features_to_plot:
    churn_rate = (
        eda_df.groupby(col)["Churn"]
        .apply(lambda x: (x == "Yes").mean())
        .sort_values(ascending=False)
    )

    churn_rate.plot(kind="bar")
    plt.title(f"Churn Rate by {col}")
    plt.ylabel("Churn Rate")
    plt.xlabel(col)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
